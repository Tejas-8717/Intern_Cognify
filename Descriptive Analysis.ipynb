```python
df['Cuisines'] = df['Cuisines'].fillna(df['Cuisines'].mode()[0])
```

```python
df.isna().sum()
```

```python
binary_cols = ['Has Table booking', 'Has Online delivery', 'Is delivering now']
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df[binary_cols].dtypes
```

```python
print(df['Aggregate rating'].describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print("\nRating Frequency Distribution:")
print(df['Aggregate rating'].value_counts().sort_index())
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['Aggregate rating'], bins=20, kde=True)
plt.title('Distribution of Aggregate Rating')
plt.xlabel('Aggregate Rating')
plt.ylabel('Count')
plt.show()

print("\nRating Text Breakdown:")
print(df['Rating text'].value_counts())
```
